2. The following function is intended to return the sum of all its arguments. Find and correct the error

```python
def calculate_sum(*args):
    total = 0
    for i in range(len(args)):
        total += args[i]
    return total

print(calculate_sum(10, 20, 30, 40))
```

3. Conceptual – append() vs extend()

What is the difference between append() and extend() in Python?

What will be the output of the following code?

```python
# append is used to add only one element in case of iterable it directly adds that to end at the last but extend is used to add iterables at the end.
a = [1, 2, 3]
a.append([4, 5])  # [1,2,3,[4,5]]

b = [1, 2, 3]
b.extend([4, 5])  # [1,2,3,4,5]

print(a)
print(b)
```

5. Find the First Non-Repeating Character

Write a function to find the first character in a string that appears only once.

```python
def non_rep(st):
  st1=set(st)
  for i in st:
   if i in st1:
      return i
st="aabbcdde"
print(non_rep(st))
```

6. Write a function that takes a list of integers and returns two separate lists: one containing even numbers and the other containing odd numbers, preserving their original order.

```python
def sep(l1):
  odd=[]
  even=[]
  for i in l1:
    if i%2==0:
      even.append(i)
    else:
      odd.append(i)
  return even, odd
l1=[11,22,33,44,55,66,77,88,99,100]
even,odd=sep(l1)
print("even num is :", even)
print('odd num is :', odd)
```

7. pallindrome search:


```python
def is_pall(s1):
  rev=''
  for i in range(len(s1)-1,-1,-1):
    rev+=s1[i]
  return s1==rev
s1="madama"
print(is_pall(s1))
```

8. Write a function to remove all occurrences of a given number from a list without using remove().

```python
def rem(l3):
  removed=set(l3)
  return removed
l3=[2,3,2,4,5,3,5,6]
print(rem(l3))


```
