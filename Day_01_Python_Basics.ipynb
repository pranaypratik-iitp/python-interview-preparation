4. Write a function to find the second-largest unique element in a list without using sort(), sorted(), or repeatedly using max().

```python
def second_largest(arr):
    largest = float('-inf')
    second = float('-inf')

    for num in arr:
        if num > largest:
            second = largest
            largest = num

        elif num > second and num != largest:
            second = num

    if second == float('-inf'):
        return None

    return second


arr = [10, 5, 8, 10, 3, 8]

print(second_largest(arr))
```

9. Write a function using *args that accepts any number of integers and returns the sum of only the even numbers. Explain what args contains inside the function.

```python
def sum_even(*args):
  sum1=0
  for i in args:
    if i%2==0:
      sum1+=i

  return sum1


print(sum_even(2,1,3,4,2,5,6,7,9,11,12,14))


# args works very different from positional arguments like in pos arg if we reserve 2 pos then only two arg is allowed but in args there is no
# limitation of arguments. syntax of args is *args or *anything
```
