1. List Comprehension: Write a list comprehension to create a list of squares of numbers from 1 to 20 that are divisible by 2 but not by 3.

```python
# List comprehension is a concise way to create a new list in Python using a single
# line of code instead of writing a traditional for loop.
# EXPRESSION FOR I IN ITERABLE. eg [x**2 for x in range(0,10)]
squares = [x**2 for x in range(1, 21) if x % 2 == 0 and x % 3 != 0]

print(squares)
```

3. Nested Lists: Given a nested list, flatten it into a single list without using libraries or recursion.

```python
nested = [[1, 2], [3, 4], [5, 6]]
flat=[]
for i in nested:
  for j in i:
    flat.append(j)
print(flat)

# by list comprehension:
flat1=[i for j in nested for i in j]
print(flat1)

# ****************** Do it by libraries too *************************
```

4. String Coding: Write a Python function to find the first character that appears more than once in a string.

```python
def first_occur(s):
  seen=set()

  for char in s:
    if char in seen:
      return char
    seen.add(char)

s='pranaypratik'
print(first_occur(s))
```

```python
# == assignment operator whereas is belongs to relational operator
a = [1, 2, 3]
b = [1, 2, 3]   # a and b are same but both belongs to different object
c = a   # both belongs to same object

print(a == b) # True
print(a is b) # False
print(a is c) # True
```

9. Remove Duplicates While Preserving Order (do it without set)

```python
def dup(l2):
  l3=[]
  for i in l2:
    if i not in l3:
      l3.append(i)

  return l3
l2=[11,12,13,12,14,11,13]
print(dup(l2))
```

10. Write a function that counts the frequency of each character in a string and returns the result as a dictionary.

```python
def char_freq(l5):
  l6={}
  for i in l5:
    if i in l6:
      l6[i]+=1
    else:
      l6[i]=1

  return l6
l5=[11,12,13,12,14,11,13]
print(char_freq(l5))
```
