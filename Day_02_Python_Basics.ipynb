3. First non-repeating character: Write a function that returns the first character that appears exactly once.

```python
def non_rep(char1):
  for i in char1:
    if char1.count(i)==1:
      return i
  return None

print(non_rep("aabbcdde"))
```

5. Move zeros: Write a function that moves all zeros to the end while maintaining the relative order of the non-zero elements.

```python
def move_zeros(l2):
  for i in l2:
    if i==0:
      l2[i]=l2[i+1]

  return l2
l2=[0, 1, 0, 3, 12]
print(move_zeros(l2))
```

6. Most frequent element: Write a function that returns the most frequent element in a list. If multiple elements have the same highest frequency, return the one that appears first.

```python
def most_freq(l1):
  count={}
  for i in l1:
    if i in count:
      count[i]+=1
    else:
      count[i]=1
  return count

l1=[11,22,22,11,23,42,33,23]
count=most_freq(l1)
max_freq=0
most_frequent=None
for i in l1:
  if count[i]>max_freq:
      max_freq=count[i]
      most_frequent=i
l1=[11,22,22,22,11,23,42,33,23]
print(most_frequent)
```
