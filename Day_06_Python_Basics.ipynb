3. Conceptual – Shallow Copy and List Slicing

How does a[:] work in Python?

Is it a shallow copy or a deep copy?

What happens when the original list contains nested lists?

```python
# a[:] copies as shallow copy in list whivh is not nested but in terms of neasted list if changes is done to one will reflect in both
# New outer list, but shared references to nested objects.
# lets dive into it with an example as:
a=[[3,4],[2,3]]
b=a[:]
print(a is b)              # False
print(a[0] is b[0])        # True
print(a[0][0] is b[0][0])  # True
```

4. Conceptual – enumerate() and zip()

What is the purpose of enumerate()?

What is the purpose of zip()?

Explain how each can simplify a for loop with a simple example.

```python
# - enamurates : whwn you are much interested in index and values both of an iterable (such as list ,tuple) then you should use enumerate


# with normal python
name=['pranay','pratik']
for i in range(len(name)):
  print(i,name[i])

# with enumerate:
for index, name in enumerate(name):
  print(index,name)


# - zip is used to combine two or more iterables:

# with python
serial=[1,2,3,4,5]
brand=['adidas','puma','h&m','reebok','zara']
for i in range(len(serial)):
  print(serial[i],brand[i])

# with zip:
for serial,brand in zip(serial,brand):
  print(serial,brand)
```

6. largest number in an array


```python
def largest(arr):
  lar=arr[0]
  for i in range(len(arr)):
    if arr[i]>lar:
      lar=arr[i]
  return lar
arr=[2,5,4,37,1,8,9,10,82,76,54]
print(largest(arr))
```

9. Find the Intersection of Two Lists

Write a function to find the common unique elements between two lists, preserving the order of their first appearance in the first list. Do not use set().



```python
def intersection_list(arr1,arr2):
  res=[]
  for i in arr1:
    if i in arr2 and i not in res:
      res.append(i)

  return res
arr1=[22,33,44,55,66,77]
arr2=[21,32,33,77]
print(intersection_list(arr1,arr2))
```
