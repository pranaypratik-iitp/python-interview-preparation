1. Write a function to find all elements that appear exactly once in a list, preserving their original order.

```python
def once(l):
  unique=set()
  for i in l:
    if i not in unique:
      unique.add(i)
  return unique

l=[11,23,45,23,56,45,56,23]
print(once(l))
```

8. Find the Missing Number

Given a list of numbers from 1 to n with one number missing, write a function to find the missing number.

```python
def find_missing(numbers):
    n = len(numbers) + 1
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(numbers)

    return expected_sum - actual_sum

numbers = [1, 2, 4, 5, 6]
print(find_missing(numbers))


# another methos - set method
def find_missing(numbers):
    num_set = set(numbers)

    for i in range(1, len(numbers) + 2):
        if i not in num_set:
            return i

numbers = [1, 2, 4, 5, 6]
print(find_missing(numbers))
```

9. Find Common Elements

Write a function to find common elements between two lists without using set(). Preserve the order of the first list and avoid duplicates in the result.

```python
def common(l1,l2):
  com=[]
  for i in l1:
    if i in l2:
      com.append(i)
  return com
l1=[11,2,3,4,22]
l2=[4,5,25,3,2]
print(common(l1,l2))
```

10. Word Frequency

Write a function to count the frequency of each word in a sentence, ignoring uppercase and lowercase differences.

```python
def freq_count(sen):
  words=sen.lower().split()
  count1={}
  for i in words:
    if i in count1:
      count1[i]+=1
    else:
      count1[i]=1
  return count1
sen="Python is easy and Python is powerful"
print(freq_count(sen))
```
